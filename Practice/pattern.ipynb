```python
lines = 5
for i in range(1, lines+1):
    for j in range(1, i+1):
        print("*",end="")
    print()
```

```python
lines = 5
for i in range(lines, 0, -1):
    for j in range(1, i+1):
        print("*",end="")
    print()
```

```python
for i in range(100, 1000):
    num = i
    temp = num
    sum = 0
    while num !=0:
        digit = num % 10
        sum += pow(digit, 3)
        num //= 10
    if temp == sum:
        print(sum)
```
